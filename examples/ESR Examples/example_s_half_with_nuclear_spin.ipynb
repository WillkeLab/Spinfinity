# ESR example: S=1/2 coupled to I=5/2

This notebook sets up a minimal spin system with hyperfine coupling and calculates the ESR spectrum using the visualization helper.

```
from spinsimulation.SpinSys import SpinSys
from spinsimulation import vis

mu_n_mev_per_t = SpinSys.muB / 1836.15267389
spinsys = SpinSys([0.5], nuclear_spins=[2.5], nuclear_magnetons=[mu_n_mev_per_t], hyperfine_A=[[0.002, 0.002, 0.012]])

spinsys.ReadoutSpin = 0
spinsys.B = [0.0, 0.0, 0.5]
spinsys.BTip = [0.001, 0.0, 0.0]
spinsys.T = 1.0

vis.plotESR_FreqSweep(spinsys, FreqRange=[6.5, 7.5], N=400, lw=0.03, norm=True)
```
