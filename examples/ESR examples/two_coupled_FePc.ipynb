# Simulating the ESR Spectrum of a two coupled FePc molecules


```python
# Import numpy
import numpy as np

# Import the Simulation packages
from spinfinity import *
%load_ext autoreload
%autoreload 2
```

```python
# Define the Spin System
spinsys=SpinSys([0.5, 0.5]) #  S=1/2

# Add the Spin Hamiltonian parameters
spinsys.B=[0,0,0.5] #Add a magnetic field in Tesla
spinsys.T=1 # Temperature

# Add Exchange and Dipole interacion
spinsys.Jvector[0,:]=np.array([0.1*spinsys.GHztomeVConversion,0.1*spinsys.GHztomeVConversion,0.1*spinsys.GHztomeVConversion])
spinsys.DipoleBool=[True]
spinsys.AtomPositions=np.array([[0,0,0],[3,4,0]])


```

## Hamiltonian Plots

```python
# Initial Calculations
hamil.calc_EigEnergies(spinsys)
hamil.calc_EigStates(spinsys)


hamil.plot_EigenMatrix(spinsys);
hamil.plot_ZeemanDiagramm(spinsys, B_array=np.linspace(0,1,100), energyscale='GHz');
```

## ESR Freq Sweeps

```python
# Compare the two ESR Simulatiions
f_array = np.linspace(13, 15, 1000)

ESR.plot_FreqSweep(spinsys, freq_array=f_array, norm=True, AllowPumping=True);
ESR.plot_FreqSweep_ESRrate(spinsys, freq_array=f_array, norm=True);
```

# Tipfield dependent Freq Sweeps

```python
ESR.plot_Tipfieldmap(spinsys, freq_array=np.linspace(13,15,1000), BTip_array=np.linspace(-0.01,0.01,1000), norm=True, AllowPumping=False);
```
