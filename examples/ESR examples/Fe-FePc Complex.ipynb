# Simulating the ESR Spectrum of a Fe-FePc complex


```python
# Import numpy
import numpy as np

# Import the Simulation packages
from spinfinity import *
%load_ext autoreload
%autoreload 2
```

```python
# Define the Spin System
spinsys=SpinSys([1,0.5])

# Basic Parameters
spinsys.B=[0,0,0.5]
spinsys.BTip=[0,0,0]
spinsys.tip='f'
spinsys.T=1

# Coupling Parameters
spinsys.Jvector=np.array([[14,14,14]])
spinsys.Dvector=np.array([1.9,0])
spinsys.AtomPositions=np.array([[0,0,0],[2,1,0]])
spinsys.DipoleBool=[True]
hamil.calc_DipolarCoupling(spinsys) # Calculate the dipolar coupling based on the positions of the atoms
hamil.calc_EigEnergies(spinsys) # Calculate the eigenenergies of the system
hamil.calc_EigStates(spinsys) # Derive the eigenstates of the system in bra ket notation
```

## Hamiltonian Plots

```python
# Plotting
hamil.plot_EigenMatrix(spinsys);
hamil.plot_EnergyVsS(spinsys);
```

## ESR Simulation

```python

ESR.plot_FreqSweep(spinsys, freq_array=np.linspace(12,16,1000), norm=True, AllowPumping=False);

# Alternative
# ESR.plot_FreqSweep_ESRrate(spinsys, freq_array=np.linspace(15,16,1000), norm=True);
#spinsys.Amp
```

## Tip field dependent ESR  

```python
BTip_values = np.linspace(-0.01, 0.01, 1000)  # Example BTip values to sweep over
ESR.plot_Tipfieldmap(spinsys, freq_array=np.linspace(13,15,1000), BTip_array=BTip_values, norm=True, AllowPumping=False, Angle=10);
```
