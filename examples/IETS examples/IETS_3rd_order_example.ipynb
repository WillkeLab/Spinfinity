```python
import matplotlib.pyplot as plt
import numpy as np

import spinfinity.dIdV3 as dIdV3
import spinfinity.dIdV as dIdV
from spinfinity.SpinSys import SpinSys

spin = SpinSys([0.5])
spin.B = [0, 0, 0.5]
spin.T = 0.1

spin.G_st = 1e-9
spin.G_ss = np.array([1e-8])
spin.G_tt = 1

spin.U = np.array([0.0])
spin.b0 = 0.0

# Dimensionless substrate exchange, not the pair-interaction Jvector.
spin.J = np.array([-0.02])

voltages = np.linspace(-2, 2, 201)

dIdV.plot_IETS(spin, V_array=voltages, norm=False, order="2nd");
dIdV.plot_IETS(spin, V_array=voltages, norm=False, order="2ndR");
dIdV.plot_IETS(spin, V_array=voltages, norm=False, order="3rd");
dIdV.plot_IETS(spin, V_array=voltages, norm=False, order="3rdR");

```
