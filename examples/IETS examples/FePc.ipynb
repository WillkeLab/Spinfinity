# Simulating IETS of a Spin 1/2

```python
# Import Python packages
import numpy as np

# Import the Simulation packages
from spinsimulation import *
%load_ext autoreload
%autoreload 2
```

```python
# Define the Spin System
spinsys=SpinSys([0.5])

# Add the Spin Hamiltonian parameters
spinsys.B=[0,0,0.5] #Add a magnetic field in Tesla
spinsys.T=1 

hamil.plotEigenMatrix(spinsys);
hamil.plotZeemanDiagramm(spinsys);
```

## IETS Simulation

```python
# dIdV Calculation

# Play with parameters
spinsys.T=1
spinsys.ReadoutSpin=0 # Set the readout spin
spinsys.TipPolarization=[0,0,0.3]
spinsys.B=[0,0,5] # Magnetic field in Tesla

hamil.calcEigEnergies(spinsys) # Calculate the Eigenenergies 
dIdV.plot_IETS(spinsys, V_array=np.linspace(-5,5,500), norm=False, order='2nd');
```
