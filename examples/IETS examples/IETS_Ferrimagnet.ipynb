# Simulating IETS Spectra of a Spin-1 Spin-1/2 ferrimagnet

```python
# Imports
import matplotlib.pyplot as plt
import numpy as np
import scipy.constants as const

from spinfinity import dIdV, SpinSys, bath, hamil


%load_ext autoreload
%autoreload 2
```

We model the FePc–Fe(C6H6) complex using two coupled spins:

- Spin 0: Fe(C6H6), S = 1.
- Spin 1: FePc, S = 1/2.

The Hamiltonian contains isotropic antiferromagnetic exchange, axial
anisotropy on the spin-1 site:

$$
H =
J\,\mathbf{S}_0\cdot\mathbf{S}_1
+ D\,S_{0z}^{2}
$$

We use J = 14 meV and D = +1.9 meV. Positive D represents easy-plane
anisotropy. Dipolar coupling and transverse anisotropy are omitted.

```python
# Define the Spin System
spin=SpinSys([1,0.5])
spin.gfactorvector = [2.0, 2.0]

# Isotropic antiferromagnetic exchange
J_meV = 14.0
hamil.assign_ExchangeInteraction(
    spin,
    [(0, 1, (J_meV, J_meV, J_meV))],
)

# Axial anisotropy on Fe(C6H6) only
D_meV = 1.9
spin.Dvector = np.array([D_meV, 0.0])

hamil.build_Hamiltonian(spin)
hamil.calc_EigEnergies(spin)
hamil.calc_EigStates(spin)

eigenstate_result = hamil.plot_EigenMatrix(spin, plot_type="p")
energy_result = hamil.plot_EnergyVsS(
    spin,
    type="z",
    energyscale="meV",
)
```

```python
# Plotting the dI/dV spectrum of both spin centers
V_array = np.linspace(-30, 30, 1000)  # Voltage range in mV

return_dict = dIdV.plot_IETS(
    spin, 
    V_array=V_array, 
    norm=False, 
    order="2ndR",  # defines the order of perturbation used
    plotAllSpins=True
)
```
