# Simulate IETS of a Fe-FePc complex

```python
# Import Python packages
import numpy as np

# Import the Simulation packages
from spinfinity import *
%load_ext autoreload
%autoreload 2
```

```python
# Define the Spin System
spinsys=SpinSys([1,0.5])

# Basic Parameters
spinsys.B=[0,0,0.5]
spinsys.BTip=[1e-9,0,0.05]
spinsys.tip='f'
spinsys.T=1

# Coupling Parameters
spinsys.Jvector=np.array([[14,14,14]])
spinsys.Dvector=np.array([1.9,0])
spinsys.AtomPositions=np.array([[0,0,0],[2,1,0]])
spinsys.DipoleBool=[True]

# Calculate the Eigenstates
hamil.calc_EigStates(spinsys)
```

## Hamiltonian Plots

```python
hamil.plot_EigenMatrix(spinsys);
hamil.plot_EnergyVsS(spinsys);
```

## IETS Spectrum

```python
dIdV.plot_IETS(spinsys, V_array=np.linspace(-30,30,1000), norm=True, order='2nd', plotAllSpins=True);
```
