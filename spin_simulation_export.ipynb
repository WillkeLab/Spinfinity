# Spin Simulation Export

This notebook reproduces the spin configuration and visualization exported from the GUI.


```python
import numpy as np
import matplotlib.pyplot as plt
from spinsimulation.SpinSys import SpinSys
import spinsimulation.SpinVisualization as vis
```

```python
spins = [1.0]
spinsys = SpinSys(spins)
spinsys.T = 1.0
spinsys.ReadoutSpin = 0
spinsys.B = [0.0, 0.0, 0.0]
spinsys.BTip = [0.0, 0.0, 0.0]
spinsys.tip = 'f'
spinsys.Dvector = np.array([0.0], dtype=float)
spinsys.Evector = np.array([0.0], dtype=float)
spinsys.U = np.array([0.0], dtype=float)
spinsys.J = np.array([0.0], dtype=float)
spinsys.V_DC = 0.0
spinsys.TipPolarization = [0.0, 0.0, 0.0]
spinsys.SamplePolarization = [0.0, 0.0, 0.0]
```

```python
# Generate visualization: Eigenstate matrix (probability)
kwargs = {'plot_type': 'p'}
result = vis.plotEigenMatrix(spinsys, **kwargs)
plt.show()
```
