# Spin 1/2 example

```python
# Import Package
from spinsimulation import *

import numpy as np
from matplotlib import pyplot as plt
```

```python
# Create Spin System
spinsys = SpinSys([0.5])

# Modify Spin Hamiltonian Parameters
spinsys.B = [0, 0, 0.5] # Tesla
```

## ESR Frequency Sweep

```python
f, I = ESR.calc_FreqSweep(spinsys, freq_array=np.linspace(13, 15, 1000));

plt.figure()
plt.plot(f, I)
plt.xlabel('Frequency (GHz)')
plt.ylabel('Intensity (a.u.)')
plt.title('ESR Spectrum')
plt.show()
```
